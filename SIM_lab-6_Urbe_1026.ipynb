# Chi-Square Goodness-of-Fit Test — Student ID 26

Class Performance Test 5 | Significance level: 5%

## Import Necessary Libraries

```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
```

## 1. Student ID's last two digits: **26**

##2: Observed Data

```python

d = 26

a = 20 + d         # C
b = 25 + (d % 10)  # C++
c = 30 + (d % 7)   # Java
e = 35 + (d % 5)   # Python

observed = np.array([a, b, c, e])

print("My ID's last two digits:", d)
print("C:", a)
print("C++:", b)
print("Java:", c)
print("Python:", e)
print("Observed (O):", observed)
print("Total responses:", np.sum(observed))
```

## 3–4. Null and Alternative Hypotheses

- **H₀:** Students have equal preferences for all four programming languages (25% each).
- **H₁:** Preferences are not equal across all four languages (at least one proportion differs).

```python

print("H0: Students have equal preferences for all four programming languages (25% each).")
print("H1: Students do not have equal preferences across all four programming languages.")
```

## 5. Expected Frequencies for Each Category

```python

total = np.sum(observed)

expected = np.full(
    len(observed),
    total / len(observed),
    dtype=float
)

print(f"Total responses: {total}")
print(f"Expected frequency per language: {total / len(observed):.2f}")
print(f"Expected frequencies: {expected}")
```

## 6. Chi-Square Statistic Calculation

Formula: $\chi^2 = \sum \frac{(O-E)^2}{E}$.

```python

chi_square_stat = np.sum(
    ((observed - expected) ** 2) / expected
)

print(f"Chi-square statistic: {chi_square_stat:.4f}")
```

## 7. Degrees of Freedom

```python

df = len(observed) - 1

print(f"Degrees of freedom: {df}")
```

## 8. Critical Value and p-Value

```python

alpha = 0.05

critical_value = stats.chi2.ppf(1 - alpha, df)
p_value = stats.chi2.sf(chi_square_stat, df)

print(f"Significance level: {alpha}")
print(f"Critical value: {critical_value:.4f}")
print(f"P-value: {p_value:.4f}")
```

## 9. Final Decision

```python

print(f"Chi-square statistic: {chi_square_stat:.4f}")
print(f"Degrees of freedom: {df}")
print(f"Critical value: {critical_value:.4f}")
print(f"P-value: {p_value:.4f}")

if chi_square_stat > critical_value:
    print("Decision: Reject H0.")
else:
    print("Decision: Fail to reject H0.")
```

## 10. Conclusion in the Scenario Context

```python

if p_value < alpha:
    print("Conclusion:")
    print("At the 5% significance level, there is sufficient")
    print("evidence that students do not have equal preferences")
    print("for C, C++, Java, and Python.")
else:
    print("Conclusion:")
    print("At the 5% significance level, there is insufficient")
    print("evidence to conclude that students have unequal")
    print("preferences for C, C++, Java, and Python.")
```

## Observed vs Expected Frequencies (Graph)

```python

languages = ["C", "C++", "Java", "Python"]
x = np.arange(len(languages))
width = 0.35

fig, ax = plt.subplots(figsize=(7, 4.5))

ax.bar(x - width / 2, observed, width, label='Observed')
ax.bar(x + width / 2, expected, width, label='Expected')

ax.set_xticks(x)
ax.set_xticklabels(languages)
ax.set_xlabel('Programming Languages')
ax.set_ylabel('Number of Students')
ax.set_title('Observed vs Expected Language Preferences (ID 26)')
ax.legend()

fig.tight_layout()
fig.savefig('Chi_Square_ID_26_Observed_vs_Expected.png', dpi=300)
plt.show()

```
