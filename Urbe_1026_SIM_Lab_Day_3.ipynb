```python
import numpy as np
from sympy import Matrix
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)

regular_wait = np.random.uniform(2, 12, 3000)
busy_wait = np.random.normal(8, 2, 3000)
busy_wait = np.clip(busy_wait, 0, None)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].hist(regular_wait, bins=30, alpha=0.5, color='blue', edgecolor='black')
axes[0].set_title('Regular Days (Uniform 2–12 min)')
axes[0].set_xlabel('Value')
axes[0].set_ylabel('Frequency')
axes[0].grid(True)

axes[1].hist(busy_wait, bins=30,alpha=0.5, color='pink', edgecolor='black')
axes[1].set_title('Busy Days (Normal, mean=8, std=2)')
axes[1].set_xlabel('Value')
axes[1].set_ylabel('Frequency')
axes[1].grid(True)
plt.tight_layout()
plt.show()




regular_mean = np.mean(regular_wait)
regular_median = np.median(regular_wait)
regular_std = np.std(regular_wait)
regular_max = np.max(regular_wait)

print("Regular Days")
print("Mean:", regular_mean)
print("Median:", regular_median)
print("Standard deviation:", regular_std)
print("Max wait time:", regular_max)


busy_mean = np.mean(busy_wait)
busy_median = np.median(busy_wait)
busy_std = np.std(busy_wait)
busy_max = np.max(busy_wait)

print("Busy Days")
print("Mean:", busy_mean)
print("Median:", busy_median)
print("Standard deviation:", busy_std)
print("Max wait time:", busy_max)
```
