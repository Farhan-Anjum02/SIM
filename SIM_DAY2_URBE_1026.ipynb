```python
import numpy as np
from sympy import Matrix
import matplotlib.pyplot as plt
import seaborn as sns
d1 = 6
d2 = 2
# Example matrix A
A = np.array([[d1+2, d2+1], [2*d1, d2+2]])

# 1. Size of a matrix
size_A = A.shape  # Returns dimensions (rows, columns)

#2. Determinant of a matrix
det_A = np.linalg.det(A)  # Returns the determinant of the matrix

 #3. Inverse of a matrix
if np.linalg.det(A) != 0:  # Check if matrix is invertible
   inv_A = np.linalg.inv(A)  # Compute the inverse of A
else:
    inv_A = "Matrix is singular, cannot compute inverse"  # No inverse if determinant is 0

# 4. Rank of a matrix
rank_A = np.linalg.matrix_rank(A)  # Returns the rank of the matrix

# 5. Eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(A)  # Compute eigenvalues and eigenvectors

# 6. Characteristic polynomial
char_poly = np.poly(A)  # Get coefficients of the characteristic polynomial

roots = np.roots(char_poly)

# Display results
print("Size of A:", size_A)
print("Determinant of A:", det_A)
print("Inverse of A:\n", inv_A)
print("Rank of A:", rank_A)

print("Eigenvalues of A:", eigenvalues)
print("Eigenvectors of A:\n", eigenvectors)
print("Characteristic polynomial coefficients:", char_poly)
print("Roots of the Characteristic Polynomial (Eigenvalues):", roots)



# Example matrix B
#change row 1, column 1
B = np.array([[d1+3, d2+1], [2*d1, d2+2]])

# 1. Size of a matrix
size_B = B.shape  # Returns dimensions (rows, columns)

#2. Determinant of a matrix
det_B = np.linalg.det(B)  # Returns the determinant of the matrix

 #3. Inverse of a matrix
if np.linalg.det(B) != 0:  # Check if matrix is invertible
   inv_B = np.linalg.inv(B)  # Compute the inverse of A
else:
    inv_B = "Matrix is singular, cannot compute inverse"  # No inverse if determinant is 0

# 4. Rank of a matrix
rank_B = np.linalg.matrix_rank(B)  # Returns the rank of the matrix

# 5. Eigenvalues and eigenvectors
eigenvalue, eigenvector = np.linalg.eig(B)  # Compute eigenvalues and eigenvectors

# 6. Characteristic polynomial
char_polys = np.poly(B)  # Get coefficients of the characteristic polynomial

root = np.roots(char_polys)



# Display results
print("Size of B:", size_B)
print("Determinant of B:", det_B)
print("Inverse of B:\n", inv_B)
print("Rank of B:", rank_B)

print("Eigenvalues of B:", eigenvalue)
print("Eigenvectors of B:\n", eigenvector)
print("Characteristic polynomial coefficients:", char_polys)
print("Roots of the Characteristic Polynomial (Eigenvalues):", root)


#1. How did the determinant change and why?
#ans. the matrix A has an negative determinant and matrix B's determinant is 0.

#2. Did the rank change? Explain.
#ans. yes. Rank of B changed to 1 while A matrix's rank is 2.

#3. How did the eigenvalues respond?
#ans. They shifted from (12.32, -0.32) to (13, 0).

#4. Is B easier or harder to invert than A? Why?
#ans. B doesn't have an inverse. A has an inverse.
```
