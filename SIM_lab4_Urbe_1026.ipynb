```python
import math
import random
import numpy as np

# Part 1: Setup & Personalization
my_id = 26  # Last two digits of Roll 1026
R = my_id / 10
full_name = "Farhan Anjum Urbe"

# Count letters in name
letter_count = 0
for char in full_name:
    if char.isalpha():
        letter_count = letter_count + 1

base_points = 200 * letter_count

print("Range Used: [0,", R, "]")
print("Number of letters in name:", letter_count)
print("Starting points:", base_points)
print("")
print("=" * 75)
print("")


# Part 2: Simulation Logic
def is_inside(x, y, rad):
    if (x * x) + (y * y) <= rad * rad:
        return True
    else:
        return False


def run_simulation(total_points, R):
    half = total_points // 2
    inside_count = 0

    # part A - using random.uniform() in a loop
    for count in range(half):
        x = random.uniform(0, R)
        y = random.uniform(0, R)
        if is_inside(x, y, R):
            inside_count = inside_count + 1

    # part B - using numpy.random.rand() arrays
    remaining = total_points - half
    x_array = np.random.rand(remaining) * R
    y_array = np.random.rand(remaining) * R
    for index in range(len(x_array)):
        x = x_array[index]
        y = y_array[index]
        if is_inside(x, y, R):
            inside_count = inside_count + 1

    return inside_count


# Part 3: Output & Analysis
print("Range Used", "Total Points", "Points Inside", "Estimated Pi", "Error")
print("-" * 75)

for i in [1, 2, 3, 4, 5]:
    total_points = base_points * i
    inside_count = run_simulation(total_points, R)
    estimated_pi = 4 * (inside_count / total_points)
    error = abs(math.pi - estimated_pi)
    print(R, total_points, inside_count, round(estimated_pi, 6), round(error, 6))
```
